## ⚙️ Configure the target

Set `TARGET` below. Use `IP` or `IP:PORT`. If you have written authorisation for this target, set `AUTHORISED = True` to enable credential testing.

```python
TARGET     = "1.2.3.4"       # e.g. "203.0.113.10" or "203.0.113.10:8081"
AUTHORISED = False           # set True ONLY if you have written permission
NO_OSINT   = False           # set True to skip Shodan/Censys/ipinfo lookups
```

## ▶️ Run the scan

Colab is non-interactive, so we pass everything on the command line and use `--yes` to auto-answer the "keep going?" prompt.

```python
import shlex
flags = ["--yes"]
if AUTHORISED: flags.append("--i-have-authorisation")
if NO_OSINT:   flags.append("--no-osint")
cmd = f"python3 CamXploit.py {shlex.quote(TARGET)} " + " ".join(flags)
print(f"$ {cmd}")
!{cmd}
```
